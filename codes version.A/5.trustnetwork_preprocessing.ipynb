```python
import csv

# 텍스트 파일에서 데이터 읽기
input_file = './dataset/trustnetwork.txt'  # 텍스트 파일 경로
output_file = 'user_link.csv'

# 텍스트 파일 열기
with open(input_file, 'r') as file:
    data = file.readlines()

# 딕셔너리 초기화
user_dict = {}

# 데이터 처리
for entry in data:
    entry = entry.strip()  # 각 줄의 양옆 공백 및 줄바꿈 제거
    user_id, trustor = entry.split("::::")[:2]
    
    # key가 이미 딕셔너리에 존재하면 set에 추가, 없으면 새로 set 생성
    if user_id not in user_dict:
        user_dict[user_id] = set()
    user_dict[user_id].add(trustor)

# CSV 파일로 저장
with open(output_file, mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["userid", "trustors"])  # 헤더 작성
    
    # 딕셔너리 내용 CSV에 저장
    for user_id, trustors in user_dict.items():
        writer.writerow([user_id, ",".join(trustors)])

print("CSV 파일이 성공적으로 저장되었습니다.")
```

```python

# 이제 CSV 파일을 다시 읽어서 set으로 변환
loaded_user_dict = {}

with open(output_file, mode='r') as file:
    reader = csv.reader(file)
    next(reader)  # 헤더를 넘김
    
    for row in reader:
        user_id = row[0]
        trustors = set(row[1].split(','))  # 쉼표로 분리하여 set으로 변환
        loaded_user_dict[user_id] = trustors

# 확인용 출력
print(loaded_user_dict)
```
